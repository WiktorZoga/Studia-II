```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from PIL import Image
```

```python
image_paths = ['bird.jpg', 'giraffe.webp', 'otter.jpg', 'panda.png']

N_COLORS = 16
```

Redukujemy palete barw do 16 kolorów

```python
for img_path in image_paths:
    img_pil = Image.open(img_path)
    img_array = np.array(img_pil, dtype=np.float64) / 255.0
    h, w, d = img_array.shape

    data = img_array.reshape((h * w, d))

    kmeans = KMeans(n_clusters=N_COLORS, random_state=2137, n_init=10)
    kmeans.fit(data)

    new_colors = kmeans.cluster_centers_
    labels = kmeans.predict(data)

    new_img_array = new_colors[labels].reshape((h, w, d))

    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img_array)
    plt.title("Oryginał (wiele kolorów)")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(new_img_array) 
    plt.title(f"K-means: {N_COLORS} Kolorów")
    plt.axis('off')

    plt.show()
```

Wydara jest szara, więc trochę średnio wyszło.

Tło pandy troszkę bardziej rozmazanę.

Reszta elegancko, dalej dobrze wygląda, chociaż charaktersytczny czerowny dziub ptaszka jest pomarańczowy.

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split

data = pd.read_csv("../data/advertising.csv")

train_data, test_data = train_test_split(data, test_size=0.2)

features = data.columns[:-1]
target = data.columns[-1]

X_train_raw, y_train = train_data[features], train_data[target]
X_test_raw, y_test = test_data[features], test_data[target]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train_raw), columns=features)
X_test_scaled = pd.DataFrame(scaler.transform(X_test_raw), columns=features)
```

```python
from sklearn.neighbors import KNeighborsRegressor
import sklearn.metrics as skm

def train_and_evaluate_knn(X_train, y_train, X_test, y_test, k, data_type):
    model_knn = KNeighborsRegressor(n_neighbors=k)
    model_knn.fit(X_train, y_train)
    y_pred = model_knn.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    return {'k': k, 'data_type': data_type, 'mse': mse, 'mae': mae, 'r2': r2}
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

k_values = [1, 3, 5, 7, 9, 11, 13, 15]
results = []

for k in k_values:
    results.append(train_and_evaluate_knn(X_train_raw, y_train, X_test_raw, y_test, k, 'raw'))
    results.append(train_and_evaluate_knn(X_train_scaled, y_train, X_test_scaled, y_test, k, 'scaled'))
```

```python
results_df = pd.DataFrame(results)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
import os
```

```python
k = 3

kmeans_model = KMeans(n_clusters=k, random_state=2137, n_init=10) # n_init=10 tzn 10 interacji
gmm_model = GaussianMixture(n_components=k, random_state=2137)

for i in range(1, 6):
    data = np.load(f"../data/data{i}.npy")
    
    # K-Means
    kmeans_labels = kmeans_model.fit_predict(data)
    kmeans_centers = kmeans_model.cluster_centers_
    
    # Gaussian Mixture Model 
    gmm_labels = gmm_model.fit_predict(data)
    gmm_means = gmm_model.means_
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    fig.suptitle(f"Zbiór danych: data{i}.npy (k=3)", fontsize=16)
    
    ax1.scatter(data[:, 0], data[:, 1], c=kmeans_labels, cmap='viridis', s=10, alpha=0.8)
    ax1.scatter(kmeans_centers[:, 0], kmeans_centers[:, 1], 
                c='red', marker='o', s=30, label='Centra K-Means')
    ax1.set_title("K-Means")
    ax1.legend()
    
    ax2.scatter(data[:, 0], data[:, 1], c=gmm_labels, cmap='viridis', s=10, alpha=0.8)
    ax2.scatter(gmm_means[:, 0], gmm_means[:, 1], 
                c='red', marker='o', s=30, label='Środki GMM')
    ax2.set_title("Gaussian Mixture Model (GMM)")
    ax2.legend()
    
    plt.show()
```

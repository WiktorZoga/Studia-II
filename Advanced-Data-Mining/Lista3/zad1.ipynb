Consider the self-attention mechanism defined as:
$$A = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right),$$

where the softmax function is applied row-wise. The matrices $Q, K ∈ \mathbb{R}^{n \times d_k}$
represent queries

and keys respectively.
1. Show that every element of the attention matrix A is strictly positive.
2. Prove that the sum of elements in each row of A is equal to 1.

3. Provide a probabilistic interpretation of the element Aij in the context of the attention me-
chanism.

# 2

i'ty wiersz macierzy A powstaje poprzez aplikace softmaxa na i'tym wierszu macierzy $\frac{1}{\sqrt{d_k}} QK^T$. Nazwijmy go $z_i$.  Wtedy $$\sum_{j=1}^{n} \text{softmax}(z_i)_{j} = \sum_{j=1}^{n} \frac{e^{z_i^j}}{\sum_{k=1}^{n} e^{z_i^k}} = \frac{\sum_{j=1}^{n} e^{z_i^j}}{\sum_{k=1}^{n} e^{z_i^k}} = 1$$

# 3
Macierze $Q$ i $K$ reprezentują zbiory zapytań i kluczy osadzonych w przestrzeni $\mathbb{R}^{d_k}$.  

Element macierzy $QK^T$ to iloczyn skalarny, który geometrycznie pełni rolę nieunormowanej miary podobieństwa między $i$-tym zapytaniem a $j$-tym kluczem.   

Duże wartości dodatnie to zbieżne kierunki, wartości ujemne to kierunki przeciwne, a bliskie zera oznaczają brak korelacji (ortogonalność).  

Funkcja softmax aplikowana na wierszach zamienia te podobieństwa w prawdopodobieństwa.

W efekcie $A_{ij}$ to waga uwagi (prawdopodobieństwo kategoryczne), która określa, w jakim stopniu $i$-te zapytanie powinien "skupić uwagę" na $j$-tym kluczu podczas budowania swojej nowej, kontekstowej reprezentacji.

Jeśli założymy, że składowe wektorów $q$ i $k$ to niezależne zmienne losowe o średniej 0 i wariancji 1, to wariancja ich iloczynu skalarnego wynosi dokładnie $d_k$ (bo sumujemy $d_k$ niezależnych iloczynów). Dlatego skalujemy prze $\sqrt{d_k}$, co daje większa stablinośc w trakcie uczenia.

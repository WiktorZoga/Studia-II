$ L^{2} = \begin{bmatrix}
    2 & -4 & 2 \\
    -4 & 14 & -10 \\
    2 & -10 & 8
\end{bmatrix}$

$x = \begin{bmatrix} 1 \\ 0 \\ 0 \end{bmatrix} $ 

$Lx = \begin{bmatrix} 1 \\ -1 \\ 0 \end{bmatrix}$

$L^{2}x = \begin{bmatrix} 2 \\ -4 \\ 2 \end{bmatrix}$


$(Lx)_{i} = \sum_{j\in N(i)} w_{ij}(x_i - x_j)$

$L^{k}x = L(L^{k-1}x) = \sum_{j\in N(i)} w_{ij}((L^{k-1}x)_i - (L^{k-1}x)_j)$

$$x^{T}Lx = \sum_{i}\sum_{j} (w_{ij} x_i^{2} - w_{ij}x_ix_j)$$

Dzięki symetrii wag

$$\sum_{i}\sum_{j} (w_{ij} x_i^{2} - w_{ij}x_ix_j) = \sum_{i}\sum_{j} (w_{ij} x_j^{2} - w_{ij}x_jx_i)$$

$$x^T L x = \frac{1}{2} \left[ \sum_{i}\sum_{j} (w_{ij} x_i^{2} - w_{ij}x_ix_j) + \sum_{i}\sum_{j} (w_{ij} x_j^{2} - w_{ij}x_ix_j) \right]$$

$$x^T L x = \frac{1}{2} \sum_{i}\sum_{j} w_{ij} (x_i^{2} - 2x_ix_j + x_j^{2})$$

$$x^T L x = \frac{1}{2} \sum_{i}\sum_{j} w_{ij} (x_i - x_j)^2$$

$$x^T L x = \sum_{(i, j)\in E} w_{ij} (x_i - x_j)^2$$
